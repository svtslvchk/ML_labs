8. Выведите минимальное и максимальное представимое значение для каждого скалярного типа numpy

````python
integer_types = [
    np.int8, np.int16, np.int32, np.int64,
    np.uint8, np.uint16, np.uint32, np.uint64
]

float_types = [
    np.float16, np.float32, np.float64
]

for dtype in integer_types:
    info = np.iinfo(dtype)
    print(f"{dtype.__name__}: min={info.min}, max={info.max}")

for dtype in float_types:
    info = np.finfo(dtype)
    print(f"{dtype.__name__}: min={info.min}, max={info.max}")
````

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
df["Systolic"] = df["Blood Pressure (s/d)"].str.split("/").str[0].astype(int)
high_stress = df[df["Stress Levels"] >= 7]
low_stress = df[df["Stress Levels"] <= 3]
high_bp = high_stress["Systolic"].mean()
low_bp = low_stress["Systolic"].mean()

print(f"Высокий стресс: {high_bp:.2f}")
print(f"Низкий стресс: {low_bp:.2f}")
````

4. Правда ли, что уровень холестерина у мужчин и женщин отличается?

````python
clstrl_by_gender = (
    df.groupby("Gender")["Cholesterol Level (mg/dL)"].agg(["mean", "median", "std", "count"])
)

print(clstrl_by_gender)
````

Уровень холестерина у мужчин и женщин практически не отличается

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
women = df[df["Gender"] == "Female"]
young = women[women["Age (years)"] < 30]
older = women[women["Age (years)"] >= 60]
young_mean = young["Cognitive Function"].mean()
older_mean = older["Cognitive Function"].mean()

print(f"Женщины до 30 лет: {young_mean:.2f}")
print(f"Женщины от 60 лет: {older_mean:.2f}")
print(f"Разница: {older_mean - young_mean:.2f}")
````

Влияет.  Наблюдается связь между активностью и весом

7. Сравните минимальный вес мужчин с высоким и низким заработком

````python
men = df[df["Gender"] == "Male"]
high_income = men[men["Income Level"] == "High"]
low_income = men[men["Income Level"] == "Low"]
min_high = high_income["Weight (kg)"].min()
min_low = low_income["Weight (kg)"].min()

print(f"Минимальный вес мужчин с высоким доходом: {min_high:.2f} кг")
print(f"Минимальный вес мужчин с низким доходом: {min_low:.2f} кг")
````

Минимальный вес мужчин с высоким доходом меньше, чем с низким доходом

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
q30 = df["BMI"].quantile(0.30)
q60 = df["BMI"].quantile(0.60)
q75 = df["BMI"].quantile(0.75)

high_bmi = df[df["BMI"] > q75]
middle_bmi = df[(df["BMI"] >= q30) &(df["BMI"] <= q60)]
high_rate = high_bmi["Chronic Diseases"].notna().mean() * 100
middle_rate = middle_bmi["Chronic Diseases"].notna().mean() * 100

print(f"BMI > 75-го перцентиля: {high_rate:.2f}%")
print(f"BMI 30–60-го перцентиля: {middle_rate:.2f}%")
````

10. Какая доля мужчин с высокой частотой пребывания на солнце имеет высокий заработок?

````python
sun_threshold = df["Sun Exposure"].quantile(0.75)
men = df[df["Gender"] == "Male"]
high_sun_men = men[men["Sun Exposure"] > sun_threshold]
high_income_rate = ((high_sun_men["Income Level"] == "High").mean() * 100)

print(f"Порог пребывания на солнце: {sun_threshold:.2f} ч/нед.")
print(f"Количество мужчин в группе: {len(high_sun_men)}")
print(f"Доля с высоким доходом: {high_income_rate:.2f}%")
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import matplotlib.pyplot as plt
import seaborn as sns
````

1. Визуализируйте частоту встречаемых хронических заболеваний

````python
diseases = df["Chronic Diseases"].value_counts()
print(diseases)

plt.figure(figsize=(9, 5))

sns.barplot(
    x=diseases.index,
    y=diseases.values,
    color="steelblue"
)

plt.title("Частота хронических заболеваний")
plt.xlabel("Заболевание")
plt.ylabel("Количество людей")
plt.grid(axis="y", alpha=0.3)

plt.show()
````

2. Как связана оценка собственных когнитивных функций с уровнем стресса?

````python
plt.figure(figsize=(9, 6))

sns.regplot(
    data=df,
    x="Stress Levels",
    y="Cognitive Function",
    scatter_kws={"alpha": 0.3, "s": 20},
    line_kws={"color": "red"}
)

plt.title("Связь когнитивных функций и уровня стресса")
plt.xlabel("Уровень стресса")
plt.ylabel("Оценка когнитивных функций")
plt.grid(alpha=0.3)

plt.show()

correlation = df["Stress Levels"].corr(df["Cognitive Function"])
print(f"Коэффициент корреляции: {correlation:.3f}")
````

3. Визуализируйте как зависит вес от роста человека

````python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Height (cm)",
    y="Weight (kg)",
    hue="Gender",
    alpha=0.5
)

plt.title("Зависимость веса от роста человека")
plt.xlabel("Рост, см")
plt.ylabel("Вес, кг")
plt.legend(title="Пол")
plt.grid(alpha=0.3)

plt.show()

correlation = df["Height (cm)"].corr(df["Weight (kg)"])
print(f"Коэффициент корреляции роста и веса: {correlation:.3f}")
````

4. Постройте box plot отображающий распределение уровня стресса у людей с различным уровнем дохода

````python
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Income Level",
    y="Stress Levels",
    order=["Low", "Medium", "High"],
    color="skyblue"
)

plt.title("Распределение уровня стресса по доходам")
plt.xlabel("Уровень дохода")
plt.ylabel("Уровень стресса")
plt.grid(axis="y", alpha=0.3)

plt.show()
````

5. Постройте круговую диаграмму отображающую как ментальное здоровье связано с паттернами сна

````python
counts = pd.crosstab(
    df["Mental Health Status"],
    df["Sleep Patterns"]
)

counts.plot.pie(
    subplots=True,
    figsize=(14, 5),
    legend=False,
    autopct="%1.1f%%",
    ylabel="",
    title=counts.columns.tolist()
)

plt.tight_layout()
plt.show()
````

7. Красиво визуализируйте как связаны между собой уровень дохода, когнитивные способности и уровень стресса*

````python
plt.figure(figsize=(11, 7))

palette = {"Low": "#E76F51", "Medium": "#E9C46A", "High": "#2A9D8F"}

sns.scatterplot(
    data=df,
    x="Stress Levels",
    y="Cognitive Function",
    hue="Income Level",
    hue_order=["Low", "Medium", "High"],
    palette=palette,
    alpha=0.35,
    s=25
)

for income in ["Low", "Medium", "High"]:
    group = df[df["Income Level"] == income]

    sns.regplot(
        data=group,
        x="Stress Levels",
        y="Cognitive Function",
        scatter=False,
        ci=None,
        color=palette[income]
    )

plt.title("Связь стресса, когнитивных функций и уровня дохода")
plt.xlabel("Уровень стресса")
plt.ylabel("Когнитивная оценка")
plt.legend(title="Уровень дохода")
plt.grid(alpha=0.2)
plt.tight_layout()

plt.show()
````
